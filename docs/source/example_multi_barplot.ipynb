The procedure follows the same logic as the {nb-download}example_single_barplot.ipynb notebook, with the only difference being that you must specify a list of annotators and their associated labels for plotting.

## Creation of a DataAplose object

```python
from pathlib import Path

import matplotlib.pyplot as plt
from pandas import Timedelta

from disclose.dataclass.detection_data import DetectionData

config = {
    "detection_file": Path(r"_static/APOCADO_results.csv"),
    "timebin_new": Timedelta("1min"),
    "filename_format": "%Y-%m-%dT%H-%M-%S",
}

data = DetectionData.from_dict(config)
```

## Barplot - All annotators / Same label

```python
fig, ax = plt.subplots(1, 1)

data.plot.histogram(
    ax=ax,
    annotator=data.annotator,
    label=data.label[0],
    bin_size=Timedelta("1h"),
)
plt.show()
```

## Barplot - One annotator / All labels

```python
fig, ax = plt.subplots(1, 1)

data.plot.set_ax(
    ax=ax,
    x_ticks_res=Timedelta("2h"),
    date_format="%H:%M",
)

data.plot.histogram(
    ax=ax,
    annotator=data.annotator[0],
    label=data.label,
    bin_size=Timedelta("1h"),
)
plt.show()
```
