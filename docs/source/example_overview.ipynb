Start with a result CSV and a small loading configuration to create a `DetectionData` object. This notebook prints the recording period, annotators, labels, and dataset names, then summarizes annotation counts by label and annotator.

```python
from pathlib import Path

from pandas import Timedelta

from disclose.dataclass.detection_data import DetectionData
```

```python
config = {
    "detection_file": Path(r"_static/APOCADO_results.csv"),
    "filename_format": "%Y-%m-%dT%H-%M-%S",
    "timebin_new": Timedelta("1min"),
}

data = DetectionData.from_dict(config)
```

Check the loaded time range and available annotators, labels, and datasets before exploring the annotation counts.

```python
print(f"{' FILE ':#^60}")
print(f"{'Begin:':<30}{data.start_datetime!s:>30}")
print(f"{'End:':<30}{data.end_datetime!s:>30}")
print(f"{'Annotators:':<30}{(', ').join(data.annotator)!s:>30}")
print(f"{'Labels:':<30}{(', ').join(data.label)!s:>30}")
print(f"{'Dataset:':<30}{(', ').join(data.dataset)!s:>30}")
```

## Overview
The overview groups the number of annotations by label and annotator. Pass a specific annotator or a list of annotators to control which rows are included.

```python
data.overview(annotator=data.annotator)
```
