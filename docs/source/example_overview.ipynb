This basic example shows you how to make an overview plot provided a YAML configuration file and a result CSV file.

```python
from pathlib import Path

from pandas import Timedelta

from disclose.dataclass.detection_data import DetectionData
```

### Load data

```python
config = {
    "detection_file": Path(r"_static/APOCADO_results.csv"),
    "filename_format": "%Y-%m-%dT%H-%M-%S",
    "timebin_new": Timedelta("1min"),
}

data = DetectionData.from_dict(config)
```

### Overview

```python
data.overview(data.annotator)
```
