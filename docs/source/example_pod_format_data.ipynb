# Process raw POD data into an APLOSE formatted DataFrame

```python
import pytz
from pandas import Timedelta
from pathlib import Path

from disclose.utils.pod import load_pod_folder, pod2aplose
```

```python
pod_files = Path(r"_static/POD_raw")
data_click = load_pod_folder(pod_files, ext="csv")

data_aplose = pod2aplose(
    df=data_click,
    tz=pytz.timezone("Etc/GMT-5"),
    dataset_name="site_name",
    label="sp_name",
    annotator="cpod",
    bin_size=Timedelta(seconds=60),
)
```
